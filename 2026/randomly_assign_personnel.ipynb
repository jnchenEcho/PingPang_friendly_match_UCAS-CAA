```python
import pandas as pd
import random

def make_teams_from_excel(
    file_path,
    output_path="随机分队结果.xlsx",
    seed=None
):
    if seed is not None:
        random.seed(seed)

    # 读取 Excel，默认第一行作为表头
    df = pd.read_excel(file_path)

    # 读取两列单位名称
    unit_A = df.columns[0]   # 国美
    unit_B = df.columns[1]   # 国科大

    # 读取人员名单，并去掉空值
    A = df[unit_A].dropna().astype(str).str.strip().tolist()
    B = df[unit_B].dropna().astype(str).str.strip().tolist()

    # 检查人数
    if len(A) != 18 or len(B) != 18:
        raise ValueError(
            f"人数不对：{unit_A} 有 {len(A)} 人，{unit_B} 有 {len(B)} 人；应各为 18 人。"
        )

    # 初始化 12 支队伍
    teams = [[] for _ in range(12)]

    # 第一步：从 A 单位随机抽 12 人，每队分 1 人
    selected_A = random.sample(A, 12)
    remaining_A = [x for x in A if x not in selected_A]

    random.shuffle(selected_A)
    for i in range(12):
        teams[i].append((unit_A, selected_A[i]))

    # 第二步：从 B 单位随机抽 12 人，每队分 1 人
    selected_B = random.sample(B, 12)
    remaining_B = [x for x in B if x not in selected_B]

    random.shuffle(selected_B)
    for i in range(12):
        teams[i].append((unit_B, selected_B[i]))

    # 第三步：A、B 剩下的 12 人随机分配，每队再加 1 人
    remaining_people = [(unit_A, name) for name in remaining_A] + [
        (unit_B, name) for name in remaining_B
    ]

    random.shuffle(remaining_people)

    for i in range(12):
        teams[i].append(remaining_people[i])

    # 整理成表格
    result = []

    for i, team in enumerate(teams, start=1):
        row = {"队伍": f"第{i}队"}

        for j, (unit, name) in enumerate(team, start=1):
            row[f"成员{j}姓名"] = name
            # row[f"成员{j}单位"] = unit

        result.append(row)

    result_df = pd.DataFrame(result)

    # 输出到 Excel
    result_df.to_excel(output_path, index=False)

    return result_df


```

```python

result = make_teams_from_excel(
    file_path="参赛人员名单.xlsx",
    output_path="随机分队结果.xlsx",
    seed=1234  
)

print(result)
```
